## Collections in Python: Foundations of Data Manipulation

A **collection** in Python is a data structure that groups multiple items into a single unit, allowing you to store and organize data effectively.

Python collections are building blocks for more complex structures, like **data frames** in libraries such as `pandas`. Lists and dictionaries, for example, can be easily converted into data frames, which allow for powerful data manipulation and analysis. Understanding Python's built-in collection structures equips you to handle data in core Python format, preparing you to seamlessly transition into advanced libraries and workflows.

---

### Lists, Tuples, Sets and Dictionaries

Python provides versatile data structures that allow you to handle data effectively:

- **Lists**: Ordered, changeable, and allow duplicates; ideal for sequences.  
  **Example**: A list like `[100, 200, 300, 400]` can represent a data frame column of monthly sales figures. Lists are mutable, so you can easily add or update values.

- **Tuples**: Ordered and unchangeable; good for fixed collections.  
  **Example**: A tuple `("Product A", 150)` can represent a product's initial data with fields like "Product Name" and "Stock." Tuples are immutable, preserving original values.

- **Sets**: Unordered, changeable, and unique; ideal for collections of distinct items.  
  **Example**: A set `{"North", "South", "East", "West"}` can validate unique regions in a data frame. Sets are mutable but don’t allow duplicates.

- **Dictionaries**: Key-value pairs, perfect for structured data.  
  **Example**: A dictionary `{"Product": "Laptop", "Price": 1200, "Stock": 50}` can represent a data frame row, with keys as column names. Dictionaries are mutable, allowing updates to specific values.


**Why Collections Matter**: Using the right collection type optimizes performance and simplifies operations like sorting, filtering, and aggregating data.

### Indexing Lists

When you create a list, Python assigns an 'index' for each value in the list.

**Positive Indexing**
The leftmost list value is assigned  an index of 0, the next value is labelled 1, the next 2, and so on until the end of the list is reached.

**Negative Indexing**
You can also count backwards from the end of the list, starting from the rightmost side with -1. 

> **Note**: Indexing applies to other collections in Python, but we will focus on practicing with lists for now.

### Slicing Lists

#### Start, Stop

Slicing can be performed accross many different data structures in Python. We will focus on slicing lists.

Slicing is a way to select a subsection of a list. 

 The general syntax for slicing is:
>list_to_slice [start : stop]

- **start** is the position to start the slice from
- **stop** is the position AFTER the end of the slice




Use list slicing to get all but the first and last values

```python
rainbow[1:5]
```

You can use a combination of positive and negative indexing for your slice.

```python
rainbow[1:-1]
```

Selecting the first three values

```python
rainbow[0:3] 
```

If you put no number before the :, the 0 is assumed. Think of this as saying 'get everything before position 3'

```python
rainbow[:3]
```

If you put no number after the :, the final index is assumed. 'get everying from position 3 onwards'

```python
rainbow[3:]
```

#### Start, Stop, Step
There is also a second colon which you can use to change the step.

    - collection[start:stop]
    - collection[start:stop:step]

```python
rainbow[1:5:2]
```

```python
rainbow[::-1]
```

You can slice a string just like a list

#### Slicing is not just for lists

```python
first_programmer = 'Ada Lovelace'
first_programmer
```

```python
first_programmer[2:7]
```

Dictionaries...

are heterogenous

are mutable

## Iteration

Sometimes we want to iterate over sequences (such as lists, tuples, strings, etc.) to perform operations on each element in the sequence.

We will look at 2 different ways of iterating over sequences in python:

- for loops
- list comprehensions


### For Loops

In Python, we use for loops to iterate through sequential data.

```python
seq = [1, 2, 3, 4, 5]
for num in seq:
    print(num)
```

```python
for num in seq:
    print('hello')
```

```python
for num in seq:
    print(num ** 2)
```

```python
sentence_words = ['The ', 'quick ', 'brown ', 'fox ', 'jumps ']
sentence = ''
for word in sentence_words:
    sentence = sentence + word
print(sentence)
```

Strings are also sequential data - think of them as being a like a list of letters.

```python
pet = 'cat'
for letter in pet:
    print(letter)
```

using range

```python
for each_num in range(6):
    print(each_num)
```

```python
for each_num in range(2,6):
    print(each_num)
```

```python
for each_num in range(3): # 0 to 10 is assumed
    print('hello')
```

### List Comprehensions

List comprehensions allow you to loop through lists in a readible, single line of code

```python
comprehension_upper_rainbow = [color.upper() for color in rainbow]
comprehension_upper_rainbow
```

list comprehensions are useful for filtering lists

```python
[color for color in rainbow if color.endswith('e')]
```

uppercase only those that end with e


```python
[color.upper() if color.endswith('e') else color for color in rainbow]
```

list comprehension to get the even numbers

```python
seq = [1, 2, 3, 4, 5]
```

```python
even_nums = [num for num in seq if num %2 == 0 ]
even_nums
```
