###  <span style="color:green">For Loops Challenge</span>

Using a for loop for each challenge:

1. Print every color in the rainbow list to a new line
2. Print every color in the rainbow to a new line, in upper case
3. Print just the first 3 letters of each color in the rainbow
4. Write a for loop that iterates over the rainbow list and prints each color along with the number of letters it contains.
5. Use a for loop to polulate a new list, uppercase_rainbow, with the colors from in rainbow in but in upper case
6. Create a new list called 'rainbow_words_reversed', which has all colors from rainbow but with each word reversed, e.g. 'red' becomes 'der'

```python
for color in rainbow:
    print(color)
```

```python
for color in rainbow:
    print(color.upper())
```

```python
for color in rainbow:
    print(f"{color} has {len(color)} letters")
```

```python
uppercase_rainbow = []

for color in rainbow:
    uppercase_rainbow.append(color.upper())
    
uppercase_rainbow
```

```python
reversed_rainbow =[]

for color in rainbow:
    reversed_rainbow.append(color[::-1])
 
reversed_rainbow
```
