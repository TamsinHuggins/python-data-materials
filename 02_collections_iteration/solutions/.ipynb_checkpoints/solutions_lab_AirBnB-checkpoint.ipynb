### 15) Write a function that will take in one whole listing and print its price. Name the function get_price. 
Call your function with the argument data[0] to check it works.

Call it a second time with the argument data[1].


```python
def get_price(listing):
    print(listing['price'])

get_price(data[0])
get_price(data[1])
```

### 16) Use a for loop to call the function get_price on the first 10 listings in the data. This should print the price of each listing.

```python
for i in range(10):
    get_price(data[i])
```

### 17) Create functionality that will recieve the name of an amenity and print out all the listings that have that amenity.

 Test your function by calling it with the argument 'Pool table'
 
 Test again your function by calling it with the argument 'Gym'


```python
def get_listings_with_amenity(amenity):
    print(f"Listings with {amenity}:")
    for listing in data:
        if amenity in listing['amenities']:
            print(listing['name'])
    

get_listings_with_amenity("Gym")
```
