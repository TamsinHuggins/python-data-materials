# Matplotlib

---

Matplotlib allows you to create publication-quality, reproducible figures: http://matplotlib.org/. Matplotlib was the first widely-used Python data visualization library, so many other libraries are built on top of it, or designed to work alongside it.

---

### High-Level APIs to Matplotlib

High-level APIs like **Pandas** and **Seaborn** simplify plotting by providing intuitive interfaces for creating common visualizations. These libraries internally use Matplotlib, meaning you can still leverage Matplotlib's customization options after generating a plot.

- **Pandas DataFrame `.plot()`**:
  - Pandas provides a direct way to create plots from DataFrames or Series.
  - Example:
    ```python
    df.plot(kind="line")  # Quickly create a line plot
    ```

- **Seaborn**:
  - A library built on top of Matplotlib for statistical data visualization.
  - It offers high-level functions like `sns.barplot()` or `sns.scatterplot()` for quickly generating attractive plots.

By starting with these high-level APIs, you can quickly visualize your data, and then drop into Matplotlib to fine-tune elements like titles, labels, or legends.

---

### Pros of Matplotlib:
- **Versatile**: Can handle almost any type of plot, from simple line charts to complex 3D visualizations.
- **Customizable**: Offers complete control over every plot element.
- **Extensive Documentation**: Plenty of guides, examples, and tutorials available.
- **Universal Tool**: Integrates well with many Python libraries and back ends.

---

### Cons of Matplotlib:
- **Verbose**: Requires more code to make each plot compared with high-level APIs (e.g., Pandas `.plot()` or Seaborn).
- **Learning Curve**: The syntax is inspired by MATLAB, which can feel unintuitive for beginners.
- **Too Flexible**: Many ways to create a single graph, leading to potential confusion.

---

### Getting the Best of Both

Matplotlib requires a lot of code to build a simple graph, but it allows total control over every element. High-level APIs like **Pandas** and **Seaborn** help you generate plots quickly but limit the amount of fine control you have. 

**The good news:** You can use a combination of both! For example:
- Generate the plot quickly with `df.plot()`.
- Then, 'drop down' into Matplotlib to adjust specific elements like ticks, labels, or legends.

---

### How Matplotlib Works in Jupyter Notebooks

In Jupyter notebooks, Matplotlib integrates seamlessly, allowing you to generate and display visualizations directly within the notebook cells. However, there are some best practices to follow:

1. **One Visualization Per Cell**:
   - Each complete Matplotlib visualization should typically go in its own cell. This ensures that the plot is rendered correctly.

2. **`plt.show()` in Jupyter**:
   - While Jupyter automatically renders the last Matplotlib object in a cell, explicitly using `plt.show()` is considered good practice.
   - It ensures:
     - The plot is displayed immediately after the relevant code.
     - Clarity when multiple plots are being developed sequentially.

## Pandas and Matplotlib

Pandas and Matplotlib are deeply integrated, with Pandas plotting methods such as `.plot()` serving as direct high-level APIs to Matplotlib.

Let's create a DataFrame to investigate.

````python
import pandas as pd


df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [3, 4, 6]
})

````

````python
df
````

````python
# Plot a single column 
df.plot()

plt.show()
````

### Pandas DataFrame `.plot()` leverages Matplotlib under the hood

e.g. what is the type of my_plot?

````python
my_plot = df.plot(title="A versus B")
type(my_plot)
````

### Equivalent code in matplotlib

````python
import matplotlib.pyplot as plt

# Create a figure and axes
fig, ax = plt.subplots()

# Plot each column in the DataFrame manually
ax.plot(df.index, df['A'], label='A')  # Line plot for column 'A'
ax.plot(df.index, df['B'], label='B')  # Line plot for column 'B'

ax.set_title('A versus B')
````

## The `figure` and `axes` are the two most important objects in matplotlib
In matplotlib, the figure and axes objects are the fundamental building blocks for creating and organizing plots.

### The figure 
Represents the entire canvas or page where your plots reside. You can think of the figure as the container for one or more axes, allowing you to arrange multiple subplots or visualizations within a single figure.

### The axes
axes represent individual plotting areas within the figure. The axes object defines the coordinate system, scales, and labels for a specific plot and provides a versatile platform for adding data, customizing visual elements, and applying various plot types.


Understanding how to manipulate figures and axes is crucial for creating tailored and complex data visualizations in matplotlib.


Create a figure, add axes and plot on them

### `plt.subplots()` 

````python
# Create a figure and 2 sets of axes 
fig, ax = plt.subplots(1, 2)
````

### ax contains an array of axes objects 

````python
ax.shape
````

### Accessing and plotting on one of the axes

````python
# Create a figure and 2 sets of axes
fig, ax = plt.subplots(1,2)

# Plot on the axes
ax[0].plot(df['B'])

plt.show()
````

### ax is iterable



````python
fig,ax = plt.subplots(1,2)

for each_ax in ax:
    each_ax.plot(df['B'])
````

## The Anatomy of a Figure


---

####  Figure
- **What it is**: The outermost container that represents the entire canvas for your plots. It can hold one or more Axes objects.  
- **How it’s used in penguins example**:  
  - Created with `fig, ax = plt.subplots(1, 2, layout="constrained")`.
  - `fig` is the Figure object that contains two Axes (`ax[0]` and `ax[1]`).  
  - The entire figure has a supertitle added using `fig.suptitle("Exploratory Analysis of Penguins Dataset", fontsize=16, weight='bold')`.  


####  Axes
- **What it is**: The individual plotting areas or subplots within the Figure. Each Axes has its own coordinate system, scales, and elements like titles and labels.  
- **How it’s used in penguins example**:  
  - Two Axes are created (`ax[0]` and `ax[1]`), each corresponding to a subplot.  
  - `ax[0]` is used for the scatter plot, and `ax[1]` is used for the histogram.  



#### X-Axis and Y-Axis
- **What it is**: The horizontal and vertical axes within an Axes that provide scales for the data.  
- **How it’s used in penguins example**:  
  - Both Axes objects (`ax[0]` and `ax[1]`) automatically generate X and Y axes.  
  - In the scatter plot (`ax[0]`), the x-axis represents `bill_length_mm`, and the y-axis represents `bill_depth_mm`.  
  - In the histogram (`ax[1]`), the x-axis represents `body_mass_g`, and the y-axis represents the frequency of body mass values.  



#### X-Axis Label and Y-Axis Label
- **What it is**: Text descriptions for the X and Y axes, indicating what data is being measured.  
- **How it’s used in penguins example**:  
  - `ax[0].set_xlabel("Bill Length (mm)")` and `ax[0].set_ylabel("Bill Depth (mm)")` adds a descriptive label of the the scatter plot axes

#### Ticks and Tick Labels
- **What it is**:  
  - **Ticks** are the marks along the axes that denote specific data values.  
  - **Tick Labels** are the numeric or categorical labels corresponding to the tick positions.  
- **How it’s used in penguins example**:  
  - The ticks and tick labels are automatically generated by Matplotlib for both subplots.  
  - You can customize them (e.g., formatting or rotation) if needed, but in this example, the defaults are used.

  
#### Title
- **What it is**: A descriptive heading summarizing the content of a specific Axes.  
- **How it’s used in penguins example**:  
  - `ax[0].set_title("Bill Length vs Bill Depth")` adds a title for the scatter plot.  
  - `ax[1].set_title("Distribution of Body Mass")` adds a title for the histogram.  



####  Figure Title
- **What it is**: A descriptive title summarizing the entire figure, placed above all the subplots.  
- **How it’s used in penguins example**:  
  - Added with `fig.suptitle("Exploratory Analysis of Penguins Dataset", fontsize=16, weight='bold')`.  
  - This title provides context for the entire visualization.



Load in penguins data

````python
df_penguins = pd.read_csv("../data/penguins.csv")
````

````python
# Create a figure and 2 sets of axes 
fig, ax = plt.subplots(1, 2, layout="constrained") # figsize=(20, 12)


# Plot 1: Scatter plot of bill length vs bill depth on the first subplot
ax[0].scatter(df_penguins['bill_length_mm'], df_penguins['bill_depth_mm'])
ax[0].set_title("Bill Length vs Bill Depth")
ax[0].set_xlabel("Bill Length (mm)")
ax[0].set_ylabel("Bill Depth (mm)")

# Plot 2: Histogram of body mass on the second subplot
ax[1].hist(df_penguins['body_mass_g'], bins=20)
ax[1].set_title("Distribution of Body Mass")
ax[1].set_xlabel("Body Mass (g)")
ax[1].set_ylabel("Frequency")

# Add a supertitle for the entire figure
fig.suptitle("Exploratory Analysis of Penguins Dataset", fontsize=16, weight='bold')


plt.show()
````

````python
# Create a figure and 2 sets of axes 
fig, ax = plt.subplots(1, 2, figsize=(11, 5), layout="constrained") 

# Plot 1: Scatter plot of bill length vs bill depth on the first subplot
ax[0].scatter(df_penguins['bill_length_mm'], df_penguins['bill_depth_mm'], alpha=0.3, color='black')
ax[0].set_title("Bill Length vs Bill Depth")
ax[0].set_xlabel("Bill Length (mm)")
ax[0].set_ylabel("Bill Depth (mm)")

# Plot 2: Histogram of body mass on the second subplot with bar outlines
ax[1].hist(df_penguins['body_mass_g'], bins=20, color='orange', alpha=0.6, edgecolor='black')
ax[1].set_title("Distribution of Body Mass")
ax[1].set_xlabel("Body Mass (g)")
ax[1].set_ylabel("Frequency")

# Add a supertitle for the entire figure
fig.suptitle("Exploratory Analysis of Penguins Dataset", fontsize=16, weight='bold')

plt.show()
````

 - **Legend**: An optional component that explains the meaning of different elements in the plot, such as lines or markers. It is especially useful when multiple data series are displayed.

### Plot multiple lines on one subplot

- **Legend**: An optional component that explains the meaning of different elements in the plot, such as lines or markers. It is especially useful when multiple data series are displayed.



````python
fig, ax = plt.subplots() 

ax.plot(df["B"], label="B")
ax.plot(df["A"], label="A")


ax.legend() # loc=1
plt.show()
````

## The Best of Both Worlds

This example demonstrates how to start with the Pandas `.plot()` API for quick visualization and enhance it using Matplotlib for finer control:

1. **Quick Plotting**:
   - The Pandas `.plot()` API is used to create a scatter plot of `carat` vs. `price` with basic styling (e.g., `color`, `alpha`, `title`, `figsize`).

2. **Matplotlib Customization**:
   - The `ax` object allows you to:
     - Adjust axis labels (`set_xlabel`, `set_ylabel`).
     - Add a grid for readability (`grid`).
     - Set axis limits to focus on specific data ranges (`set_xlim`, `set_ylim`).

3. **Flexibility**:
   - This approach combines the simplicity of Pandas for plotting with the detailed control of Matplotlib for customization.


````python
import matplotlib.pyplot as plt

# Create a scatter plot using the Pandas API
ax = df.plot(
    kind='scatter',
    x='carat',
    y='price',
    title='Carat vs. Price',
    color='purple',
    alpha=0.3,
    figsize=(10, 6)
)

# Use Matplotlib to further customize the plot
ax.set_xlabel("Carat Weight")  # Customize the x-axis label
ax.set_ylabel("Price (USD)")   # Customize the y-axis label
ax.grid(True, linestyle='--', alpha=0.6)  # Add a grid with custom style
ax.set_xlim(0, 4)  # Set the range for the x-axis
ax.set_ylim(0, 20000)  # Set the range for the y-axis


# Show the plot
plt.show()
````
