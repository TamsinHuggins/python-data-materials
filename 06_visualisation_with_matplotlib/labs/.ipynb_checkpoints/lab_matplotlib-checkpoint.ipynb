# Lab: Matplotlib
In this exercise, you will create stock price visualizations featuring the daily close prices of Microsoft (MSFT) and Apple (AAPL).

You will visualise the same data in **two different ways**.

This challenge is designed to reinforce your understanding of Matplotlib's **figure** and **axes** objects, which are the fundamental building blocks for creating plots. By completing this task, you will:

- Know how to create and customize subplots using the `figure` and `axes` objects.
- Practice using `axes` methods to add titles, labels, and legends to your visualizations.
- Gain hands-on experience in visualizing data using matplotlib, exerting fine control over the features within a figure.


## 1. Load and Prepare the Data
Load the `close_prices.csv` file located in the data directory into a Pandas DataFrame. Set the `date` column to be the index. Verify that the index has been successfully converted to a `DatetimeIndex`.


## 2. One Plot, Two Lines

Create a stock price chart using Matplotlib that displays the daily close prices of Microsoft (MSFT) and Apple (AAPL). Lines representing each stock should be plotted on the **same set of axes**.

- Use the provided `close_prices_df` DataFrame, which contains daily close prices for Microsoft (MSFT) and Apple (AAPL).
- Create a figure with **one set of axes**:
  - Plot **Microsoft's daily close prices** with the ticker symbol "MSFT" in **blue**.
  - Plot **Apple's daily close prices** with the ticker symbol "AAPL" in **orange**.
- Customize the chart:
  - Add an **x-axis label** ("Date") and a **y-axis label** ("Price (USD)").
  - Add a **title** summarizing the chart:  
    **"Daily Close Prices: Microsoft (MSFT) and Apple (AAPL) (2024)"**
  - Include a **legend** to identify each stock.


- Save the completed figure to the output folder: call it "one_plot_two_lines"

## 3. Two Subplots, One Figure
Create a stock price chart using Matplotlib that displays the daily close prices of Microsoft (MSFT) and Apple (AAPL) on **separate subplots**.
- Use the provided `close_prices_df` DataFrame, which contains daily close prices for Microsoft (MSFT) and Apple (AAPL).
- Create a figure with two vertically stacked subplots (**2 x 1 layout**):
  - **Top subplot**: Plot Microsoft's daily close prices with the ticker symbol "MSFT."
  - **Bottom subplot**: Plot Apple's daily close prices in orange with the ticker symbol "AAPL."
- Customize the chart:
  - Add **x-axis labels** ("Date") and **y-axis labels** ("Price (USD)") to both subplots.
  - Add clear **titles** to each subplot, specifying the company and ticker symbol.
  - Include **legends** to identify each stock.
- Add an **overall figure title** summarizing the chart:  
  **"Daily Close Prices: Microsoft (MSFT) and Apple (AAPL) (2024)"**

- Save the completed figure to the output folder: call it "two_subplots"
