In this exercise, you will create stock price visualizations featuring the daily close prices of Microsoft (MSFT) and Apple (AAPL).

- Practice using `axes` methods to add titles, labels, and legends to your visualizations.

## 1. Load and Prepare the Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
close_prices_df = pd.read_csv("../data/close_prices.csv", index_col="date", parse_dates=True)
close_prices_df
```

## 2. One Plot, Two Lines


```python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(20, 12), layout="constrained")

# Plot Microsoft
ax.plot(close_prices_df['MSFT'], label="Microsoft (MSFT)")
# Plot Apple
ax.plot(close_prices_df['AAPL'], label="Apple (AAPL)", color="orange")

# Set labels, title, and format the x-axis
ax.set_xlabel("Date")
ax.set_ylabel("Price (USD)")


# Add a legend to the plot
ax.legend()

# Overall figure title
fig.suptitle("Daily Close Prices: Microsoft (MSFT) and Apple (AAPL) (2024)", fontsize=16)

# Save and show the figure
plt.savefig("../example_output/one_plot_two_lines")
plt.show()
```

## 3. Two Subplots, One Figure

```python
import matplotlib.dates as mdates

fig, ax = plt.subplots(2, 1, figsize=(20, 12), layout="constrained")

# Microsoft subplot
ax[0].plot(close_prices_df['MSFT'], label="MSFT")
ax[0].set_xlabel("Date")  # Standard x-axis label for time series
ax[0].set_ylabel("Price (USD)")  # Clarify the unit of the y-axis
ax[0].set_title("Microsoft (MSFT)")  # Add ticker symbol to subplot title

# Apple subplot
ax[1].plot(close_prices_df['AAPL'], label="AAPL", color="orange")
ax[1].set_xlabel("Date")  # Standard x-axis label for time series
ax[1].set_ylabel("Price (USD)")  # Clarify the unit of the y-axis
ax[1].set_title("Apple (AAPL)")  # Add ticker symbol to subplot title


# Overall figure title
fig.suptitle("Daily Close Prices: Microsoft (MSFT) and Apple (AAPL) (2024)", fontsize=16)

plt.savefig("../example_output/two_subplots")
plt.show()
```
