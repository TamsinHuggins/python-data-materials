## Getting Started with Jupyter

A **Jupyter Notebook** is a web application that allows you to create and share documents that contain code, annotated with text, visualisations and equations.

A notebook can have any number of cells.

## Logic and Flow Control

### Conditionals

We often want to test if a condition is True or False.

There are a number of conditional operators in Python:
- **==** : Check if two elements are equal
- **!=** : NOT Equal
- **>**  : Greater than
- **>=** : Greater than OR equal to
- **<**  : Less than
- **<=** : Less than OR equal to

    >Notice the difference between "a single equals" and "double equals"
    >- A single equals makes a variable equal to something (assignment)
    >- Double equals is "asking" if two statements are the same (querying)


Some conditional checks

```python
x = 10
y = 12

# Output: x == y is False
print('x == y is', x == y)

# Output: x != y is True
print('x != y is', x != y)

# Output: x > y is False
print('x > y  is', x > y)

# Output: x < y is True
print('x < y  is', x < y)

# Output: x >= y is False
print('x >= y is', x >= y)

# Output: x <= y is True
print('x <= y is', x <= y)
```

### if with no else

- Code in the **if** block is executed **if** conditional statement evaluates to true. 
- Otherwise, nothing happens


```python
num = -1

# Test is num greater than 0
if num > 0:
    # Num is less than 0 so this part should NOT run
    print(num, "is a positive number.")
    print("This is only printed if num > 0.")
```

### if/else

- Code in the **if** block is executed **if** conditional statement evaluates to true. 
- if the conditional evaluates to false, code in the **else** block is executed.

```python
num = -1

if num > 0:
    print(num, "is a positive number.")
else:
    print(num, "is a negative number or 0")
```

### elif

- first, the conditional statement in the **if** block is evaluated.
    - if true, code in the if block is executed
    - if false, conditional elif statement is evaluated
- if the elif statement evaluates to true, code in the **elif** block is executed
- in cases where no **if** or **elif** statements evaluate to true, the **else** block is executed

```python
num = -1

if num > 0:
    print(num, "is a positive number.")    
elif num == 0:
    print(num, "is zero")
else:
    print(num, "is a negative number.")
```

#### Testing if a value is "in" a collection

- in
- not in

e.g.
- Check if the letter 'a' is in a list:
    - **'a' in my_list**
- Check if the key 'b' is in a dictionary:
    - **'b' in my_dict**

```python
rates = {'EUR': 1.09, 'USD': 1.28, 'JPY': 99.25}

# check if the key 'JPY' is in the dictionary 'rates'
if 'JPY' in rates:
    print('JPY rate is:', rates['JPY'])
else:
    print('not found in dictionary')
```

### Nested if/else Statements

We will explore using nested if/else statements to handle user input

User input will enter the name and age of their pet

```python

age = int(input("Enter your pet's age in years: "))
pet_type = input("Enter the type of pet [Dog/Cat]: ").lower()
```

nested if /else

### Pet identifier (nested if/else)

```python

# Input details about the pet
age = int(input("Enter your pet's age in years: "))
pet_type = input("Enter the type of pet [Dog/Cat]: ").lower()


# Determine the stage of the pet based on its age
if age < 2:
    if pet_type == "dog":
        print("Puppy")
    else:
        print("Kitten")
else:
    if pet_type == "dog":
        print("Adult Dog")
    else:
        print("Adult Cat")
```
