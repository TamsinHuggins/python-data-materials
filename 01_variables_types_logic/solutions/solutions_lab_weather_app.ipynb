# Weather Advisory Challenge

---

In this challenge, you’ll build a simple weather advisory program that gives clothing recommendations based on the weather.

#### The data
You’ll use three pieces of data to simulate weather conditions:
- `temperature` (an integer, representing temperature in °C)
- `condition` (a string, such as `"sunny"` or `"rainy"`)
- `wind_speed` (an integer, representing wind speed in km/h)

#### The task
Write a program that prints out clothing advice based on these variables.

#### The behavior
- If the temperature is above 35°C, the program should print: `"It's too hot. Don't go outside at all."`
- If it’s rainy:
  - If the temperature is above 10°C, print: `"It's rainy and over 10°C. Wear a light coat."`
  - If the temperature is 9°C or lower, print: `"It's rainy and under 9°C. Wear a warm coat."`
- If it’s sunny:
  - If the temperature is over 20°C, print: `"It's sunny and over 20°C. Wear sunglasses and cool clothes."`
  - Otherwise, print: `"It's sunny. Wear sunglasses."`
- For any other weather condition, print: `"Dress comfortably for the weather."`
- If the wind speed is above 30 km/h, add this line to the output: `"It's windy. Consider a windbreaker or an extra layer."`

#### Example outputs
Your program should print something like:
- "It's rainy and over 10°C. Wear a light coat."
- "It's sunny and over 20°C. Wear sunglasses and cool clothes."
- "It's windy. Consider a windbreaker or an extra layer."


```python
# Define the weather data for the day
temperature = 15  # example temperature
condition = "rainy"  # example condition
wind_speed = 20  # example wind speed in km/h

# Main condition based on temperature
if temperature > 35:
    print("It's too hot. Don't go outside at all.")
else:
    # Nested condition for rainy or sunny weather
    if condition == "rainy":
        if temperature >= 10:
            print("It's rainy and over 10°C. Wear a light coat.")
        else:
            print("It's rainy and under 9°C. Wear a warm coat.")
    elif condition == "sunny":
        if temperature > 20:
            print("It's sunny and over 20°C. Wear sunglasses and cool clothes.")
        else:
            print("It's sunny. Wear sunglasses.")
    else:
        print("Dress comfortably for the weather.")

    # Additional condition based on wind speed
    if wind_speed > 30:
        print("It's windy. Consider a windbreaker or an extra layer.")
```
