# Logic Flow Control Challenge:  Mission Orion Game

```python

print("Welcome to Mission Orion. Your mission is to retrieve a space probe lost in the asteroid belt.")

choice1 = input('Will you direct the mission towards "Asteroid-Belt", "Ceres", or "Phobos"? Type your destination \n').lower()
if choice1 in ("asteroid-belt", "ceres"):
    choice2 = input("An asteroid is on a collision course with your spacecraft. How many of the ship's 10 thrusters do you engage?\n")
    if int(choice2) < 3:
        print("The ship does not accelerate enough. The asteroid hits the ship and you lose control. Game Over.")
    elif 3 <= int(choice2) < 8:
        choice3 = input("You avoid the asteroid. Ahead, you detect 2 objects: one emitting a weak radio signal and one reflective metallic object. Which object do you investigate? Type 'radio' or 'metallic'. \n").lower()
        if choice3 == "radio":
            print("The object is the lost probe! You successfully retrieve it and complete your mission. You Win!")
        elif choice3 == "metallic":
            print("It’s just a shiny chunk of asteroid. Ground control recalls the mission. Game Over.")
        else:
            print("You chose an object that does not exist. Your crew question your judgement. Mutiny ensues. Game Over.")
    else:
        print("Uncontrolled acceleration sends you veering into another asteroid. Game Over.")
elif choice1 =="phobos":
    print("You get caught in Phobos's gravitational pull and crash. Game Over.")
else:
    print(f"{choice1.capitalize()} isn't a valid destination. Game Over.")
```
