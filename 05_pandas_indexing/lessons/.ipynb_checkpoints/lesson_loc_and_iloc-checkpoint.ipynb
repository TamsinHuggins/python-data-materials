
# Selecting Data with `loc` and `iloc`
---

In this notebook, we’ll focus on selecting and slicing data in pandas DataFrames using two powerful indexing methods: `loc` and `iloc`. Mastering these tools is essential for efficient data manipulation and analysis in Python.

The index in a DataFrame is a label-based reference for rows. It often serves as a unique identifier for each row, which can be leveraged with `loc` for quick access and more intuitive data manipulation.

### What You’ll Learn
- **`loc`**: Indexing by label, allowing you to select rows and columns based on names and conditions.
- **`iloc`**: Indexing by position, enabling you to select rows and columns using integer positions.
- **`set_index`**: How to set a column  to be the index.
- **DataFrame Slicing**: Practice selecting specific subsets of data.


Import pandas and load the data

```python
import pandas as pd
df = pd.read_csv('../data/world_population.csv')
df
```

---

## `iloc`

The `iloc` indexer is used to select rows and columns in a DataFrame based on their position (i.e., by numerical index). This makes it very useful when you want to select data from a specific location in the DataFrame.

### Syntax

For selecting a row based on its position:

> df.iloc[row_index]

For selecting a piece of data based on row and column position

>df.iloc[row_index, column_index]



Note: Why Use Square Brackets?
Unlike some other DataFrame operations, iloc is not a method, so we use square brackets [] instead of round brackets () to access it. This is because iloc is an indexer, a tool for accessing specific parts of the data. Square brackets signal that we’re selecting a specific slice or subset of data, much like indexing or slicing with lists in Python.

Select the row at position 210

```python
df.iloc[210]
```

select the row at position 23

select the item at row 210 and column 4

```python
df.iloc[210, 4]
```

## Setting a New Index with `set_index`

To make the 'CCA3'column the index of the DataFrame, use the `set_index` method. This allows you to access rows directly by their 'CCA3' values instead of the default integer index.

### Why Set a Custom Index?

By setting CCA3 as the index, you enable more intuitive data selection using the loc indexer. Instead of selecting rows by integer position, you can now use loc to select rows based on the country codes in the CCA3 colu

### What can be used as an index?
Commonly, a specific column is set as the index, especially if it contains identifiers or labels that represent rows uniquely, such as IDs, dates, or categories.

### Does an index have to be unique for every row?
**Uniqueness:** An index doesn’t have to be unique, but it’s often beneficial for easier data retrieval and analysis. For example, if each row represents a unique entry (like a country code or product ID), then a unique index makes accessing specific rows with loc more predictable and reliable.

**Non-Unique Indexes:** You can have non-unique index values, which is useful in cases like grouping by categories. When using loc with a non-unique index, it will return all matching rows.




```python
df.set_index('CCA3', inplace = True)
df
```

## Using `loc`

Use `loc` to select the row representing France (CCA3 value is FRA)

```python
df.loc['FRA']
```

Use `loc` to select the row representing Ireland (CCA3 value is IRL)

```python
df.loc['IRL']
```


---
## Slicing with `iloc`

Slicing with `iloc` allows you to select specific portions of your DataFrame based on numerical indices. This is extremely useful when you want to:

- **Inspect Data**: View a subset of your data to understand its structure or contents.
- **Subset Data**: Create smaller DataFrames for analysis or testing.
- **Manipulate Data**: Apply operations to specific rows or columns.

### Syntax 
> df.iloc[ start_row : end_row , start_column : end_column ]


### Notes

- **Indexing is zero-based**: The first element is at position 0.
- **The end index is exclusive**: The element at the end index is not included in the slice.



Select the first 5 rows

```python
df.iloc[0:5] 
```

Select the first 5 rows, with 0 assumed for start_row

```python
df.iloc[:5] 
```

Select rows from position 5 to 7, columns from position 6 to 9

```python
df.iloc[5:8, 6:10]
```

Select all rows for columns at positions 1 to 3

```python
df.iloc[:, 1:4]  
```
