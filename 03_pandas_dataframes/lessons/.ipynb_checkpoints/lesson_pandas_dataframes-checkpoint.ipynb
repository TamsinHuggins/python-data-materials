## Conditionals

We often want to test if something is <span style="color:blue">True </span> or <span style="color:red"> False. </span> For this we use conditional statements or "conditionals".

There are a number of conditional operators in Python:
- **==** : Check if two elements are equal
- **!=** : NOT Equal
- **>**  : Greater than
- **>=** : Greater than OR equal to
- **<**  : Less than
- **<=** : Less than OR equal to

    >Notice the difference between "a single equals" and "double equals"
    >- A single equals makes a variable equal to something (assignment)
    >- Double equals is "asking" if two statements are the same (querying)
    

## Using conditionals to filter

Create a condition that says the number of bedrooms must be greater than 20.

```python
condition = df["bedrooms"] > 20
condition
```

Use square brackets to access the part of the DataFrame in which that condition evaluates to True.

```python
df[condition]
```

find the price of the hosus with the most bedrooms

```python
df[condition]["price"]
```

 ## Selecting a single value

That gives us unwanted Pandas.Series data, so call the in-built .item() method to get just the value.

```python
df[condition]["price"].item()
```

```python
print("The price of the house with the most bedrooms is:", df[condition]["price"].item() )
```

## Creating a new column based on a calculation


```python
conversion_rate = 0.95

df['price_in_euros'] = df['price'] * 0.95
```

```python
df
```

If we care about the position where our column is placed into the data, we can use an insert statement

df.insert(loc = location for column, column= the column name, value = value(s) for the column)

```python
df.insert(loc=3, column='price_in_euros_insert', value=df['price'] * 0.95 )
```
