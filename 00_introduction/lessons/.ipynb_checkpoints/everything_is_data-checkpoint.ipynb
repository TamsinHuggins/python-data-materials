## Python for Semi-Structured Data

**Semi-Structured Data**: Semi-structured data does not conform to a rigid structure like structured data but still contains tags or markers to separate data elements. JSON (JavaScript Object Notation) is a common format for semi-structured data.

### Why Use Python for JSON?

- **Ease of Use**: Python's `json` module makes it straightforward to parse and manipulate JSON data.
- **Integration**: Python can easily integrate JSON data with other data processing libraries like `pandas` for further analysis.
- **Flexibility**: Python can handle nested and complex JSON structures, making it versatile for various applications.

### JSON and the Web

JSON is prevalent over the internet as it is the standard format for data exchange between web servers and clients. APIs (Application Programming Interfaces) commonly use JSON to send and receive data, making it essential for web development and data analysis.

### Web Scraping

Web scraping involves extracting data from websites, and Python is particularly useful for this task due to libraries like `BeautifulSoup`. These libraries can parse HTML and XML documents, allowing you to extract and manipulate data from web pages, which is often in JSON format.

Python's capabilities make it an excellent choice for working with semi-structured data like JSON.

```python
import json
with open('../data/airbnb.json', 'r',encoding='utf-8') as f:
    data = json.load(f)
```

What does one listing look like in raw JSON form?

```python
data[0]
```

Very simple filter search tool for finding homes that have a certain amenity

```python
def get_listings_with_amenity(amenity):
    print(f"Listings with {amenity}:")
    for listing in data:
        if amenity in listing['amenities']:
            print(listing['name'])
    
amenity = input('Which amenity are you interested in?')
get_listings_with_amenity(amenity)
```

### Python for Semi-Structured Data Conclusion

- Scrape data from websites to gather data
- Process data in its original JSON format without needing to force it into a structured format, preserving compatibility with semi-structured data like nested or hierarchical structures.
 

## Python for Image Data

**Image Data**: Image data consists of pixels arranged in a grid, where each pixel represents a color. Processing image data involves tasks like image recognition, classification, and transformation.

**Why Python for Image Data?**

- **Image Processing Libraries**: Python offers powerful libraries like `OpenCV`, `PIL (Pillow)`, and `scikit-image` for image processing tasks.
- **Deep Learning Frameworks**: Libraries such as TensorFlow, Keras, and PyTorch provide tools for building and training deep learning models for image recognition and classification.
- **Community and Resources**: A large community and extensive resources, including tutorials and pre-trained models, make it easier to get started with image processing in Python.

Python's extensive libraries and tools make it an excellent choice for handling and processing image data.

```python
import os
import numpy as np
from scipy import ndimage
import matplotlib.pyplot as plt
import imageio.v2 as imageio 

def generate_sketch(input_path, sigma=2, threshold=7, black_threshold=210):
    # Load the image
    image = imageio.imread(input_path, mode='F')
    
    # Apply Gaussian filter with specified sigma for smoothing
    smoothed_image = ndimage.gaussian_filter(image, sigma=sigma)
    
    # Apply Sobel filter in both x and y directions, and combine
    sobel_combined = np.hypot(ndimage.sobel(smoothed_image, axis=0), ndimage.sobel(smoothed_image, axis=1))
    
    # Apply a threshold to enhance edges
    sobel_combined[sobel_combined < threshold] = 0
    
    # Invert the image to get a sketch effect
    sketch = 255 - sobel_combined  # Inverting the colors
    
    # Turn shades of grey past the black_threshold to black
    sketch[sketch < black_threshold] = 0
    
    # Display the original and sketch images
    plt.figure(figsize=(10, 5))
    
    # Show original image
    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap='gray')
    plt.title("Original Image")
    plt.axis('off')
    
    # Show sketch
    plt.subplot(1, 2, 2)
    plt.imshow(sketch, cmap='gray')
    plt.title("Sketch Effect")
    plt.axis('off')
    
    plt.show()

# Call the function to create sketch and display images
generate_sketch(input_path='../data/spaniel.jpg')
```

## Python for Unstructured Data, e.g. Natural Language Textual Data

**Textual Data**: Textual data, like tweets and social media posts, are unstructured because they lack a predefined format. Unlike structured data in rows and columns, unstructured data is free-form text, varying in length and content, making it harder to process with traditional tools.

**Why Python for Textual Data?**

- **NLP Libraries**: Python offers libraries like `NLTK` for text processing.
- **Flexibility**: Easily manipulate and transform text for tasks like cleaning and sentiment analysis.
- **Integration**: Works well with other data processing and machine learning libraries.
- **Temporal Dependencies**: Python has tools that help work with the order of words in text data. For example, understanding the sequence of words in a sentence to determine its meaning. 

Python's libraries and tools make it ideal for handling unstructured textual data.
