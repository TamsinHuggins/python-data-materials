## Starter code explanation

The `Employee` class is an example of **Object-Oriented Programming (OOP)** in Python. Here's a breakdown of its components:

### Class Definition
- **Class Name**: `Employee`
- This class serves as a blueprint for creating `Employee` objects.

### Attributes
- **Attributes**: `name`, `salary`, `joined`
  - `name`: Stores the name of the employee.
  - `salary`: Stores the salary of the employee.
  - `joined`: Stores the date and time of when an employee object is created. This is automatically set to the current date and time by using `datetime.now()`

### Methods
- **Methods**: `__init__`, `year_joined`
  - `__init__`: A special method called a constructor, used to initialize the new `Employee` object with `name`, `salary`, and `joined` attributes.
  - `year_joined`: Returns the year the employee joined, extracted from the `joined` attribute.

This encapsulation of data (attributes) and methods (functions) into a programming entity (class) is central to the concept of OOP, which aims to combine related properties and behaviors into individual objects.


```python
from datetime import datetime

class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        self.joined = datetime.now()

    def year_joined(self):
        return self.joined.year
```


## 2. Expand the Employee class.
Extend the Employee class to manage employees more effectively. You will add several new features to the class including managing employment status, updating salary, and providing a formatted string output of employee details.


#### Add Attributes:
- **`employment_status`**: Employees should have an `employment_status` attribute that can be set to one of the following values: "full-time", "part-time", "contract". In the constructor (init) function, give this employment status attribute a default value of "full-time"

#### Implement Methods:
- **`pay_raise`**: Implement a method called pay_raise that increases an employee’s salary by a specified amount.
- **`month_joined`**: Create a method called month_joined that extracts the month the employee joined, and returns it.
- **`summary`**: Write a method called summary that returns a string containing the employee's name, salary, and join date (month/year). An example result of calling `.summary()` could be'Priya Smith earns $100000, joined 12/2024, and works full-time'

#### Test Your Class:
- Create several `Employee` objects with various attributes.
- Test that Employees initialised without an argument for employment_status default to having full-time employment
- Test your summary function prints out a summary of the employee
- Give some employees a pay raise and print out their details to check if all changes are reflected accurately.

**A Specific Test Case**
- Initialise an employee in the following way:
`priya = Employee("Priya Smith", 90000)`
- Give priya a 5k raise by calling
`priya.pay_raise(5000)`
- Call the `.summary()` function on priya. You should see something very similar to:
'Priya Smith earns $95000, joined 12/2024, and works full-time'

**Expected Behavior**
- The pay raise method should correctly add the specified amount to the current salary.
- The string representation method should neatly format all current details of an employee, making it easy to print and read.
- The pay raise should update the salary attribute of the employee to add the amount specified; this should also be reflected when `.summary()` is called

This project will help you understand class design, methods, encapsulation, and simple data validation in Python. It's a hands-on way to learn how to structure your code in a real-world scenario.


```python
from datetime import datetime

class Employee:

    def __init__(self, name, salary, employment_status="full-time"):
        self.name = name
        self.salary = salary
        self.joined = datetime.now()
        self.employment_status = employment_status
        
    def year_joined(self):
        return self.joined.year

    def month_joined(self):
        return self.joined.month
    
    def pay_raise(self, amount):
        self.salary += amount

    def summary(self):
        return f"{self.name} earns ${self.salary}, joined {self.month_joined()}/{self.year_joined()}, and works {self.employment_status}"
```

```python
priya = Employee("Priya Smith", 90000)
```

```python
priya.pay_raise(5000)
```

```python
priya.summary()
```
