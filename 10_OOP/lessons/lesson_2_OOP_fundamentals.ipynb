# Object Oriented Programming (OOP) - Fundamentals
Object-Oriented Programming (OOP) is a programming paradigm based on the concept of **objects**, which can contain **data** and **code**: data in the form of fields (often known as **attributes** or properties); and code in the form of procedures (often known as **methods**). This approach allows you to structure your software in a way that is extensible, maintainable, and intuitive, by encapsulating data and the operations on data in units called **classes** and **objects**. This encapsulation facilitates not just code organization and reuse but also the creation of complex data models that reflect real-world scenarios.

Key Concepts of OOP in Python:
 - **Classes**:
   - Definition: A class is a blueprint for creating objects. It defines a datatype by bundling data and functionalities together.
   - Example: Consider a class `Dog`. It represents the concept of a dog, not any specific dog.
 - **Objects**:
   - Definition: An object is an instance of a class. When class is defined, no memory is allocated until an object of that class is created.
   - Example: `my_dog = Dog()` creates an object `my_dog` of the class `Dog`.
 - **Attributes**:
   - Definition: Attributes are **data** stored inside a class or instance and represent the **state** or quality of the class or instance.
   - Types:
     - Instance Attributes: Unique to each instance (`my_dog.name = 'Buddy'`)
     - Class Attributes: Shared by all instances of a class (`Dog.species = 'Canis familiaris'`)
 - **Methods**:
   - Definition: Methods are **functions** defined inside a class. They are used to define the **behaviors** of an object.
   - Example: A method `bark()` in the `Dog` class might make a dog object ‘speak’.

#### Setting class attributes
Class attributes in Python are variables defined at the class level. They are shared by all instances of the class.

```python
class Shape:
    name = 'shape'
```

```python
sh = Shape()

sh.name
```

#### Setting instance attributes with `__init__()`
When the constructor method is invoked (by using parentheses after the name of the class), two things happen under the hood. Firstly, python calls the `__new__()` method of the class. This creates a new object with a location in memory (its `id`). Secondly, python passes this object to the `__init__()` or initialiser method of the class. This method can initialise the new object by setting additional **instance variables**.

The first argument to this `__init__()` method (the new object itself) is passed implicitly. It still requires a name however. This will act as a handle to refer to the object within its methods (including this one). This name can be anything but the strong convention is to use `self`. Any subsequent parameters defined in the signature of `__init__()` are used to initialise instance attributes.

```python
class Shape:
    name = 'shape'
    def __init__(self, n_sides):
        self.n_sides = n_sides
```

```python
# sh = Shape()
```

```python
tri = Shape(3)
```

```python
tri.name, tri.n_sides
```

#### Methods
Methods in Python are functions defined within a class that operate on the instances of the class. They are an essential part of object-oriented programming in Python, allowing objects to have **behaviors** in addition to attributes. Methods encapsulate behavior that is relevant to the class, and can interact with the class's attributes (state) and other methods, encapsulating functionality within the class.

```python
class Shape:
    name = 'shape'
    def __init__(self, n_sides):
        self.n_sides = n_sides
        
    def print_sides(self):
        print(f'This {self.name} has {self.n_sides} sides.')
```

```python
tri = Shape(3)
```

```python
tri.print_sides()
```

# Inheritance
Inheritance is a fundamental concept of object-oriented programming. It allows a new class to inherit attributes and methods from an existing class. This mechanism promotes code reuse and establishes a hierarchical relationship between classes.

When a class inherits from another class, it automatically takes on the attributes and methods of the parent class. This allows the child class to use and, optionally, override or extend the functionality of the parent class if it requires different functionality. Overriding methods of the parent class is done by defining a method with the same name in the child class. Extending functionality is ahchieved by adding new methods or attributes.

### Create a child class to represent a `Rectangle`
To create a child or derived class from a base class, we include the name of the base class in parentheses after declaring the name of the new class. If we don't add any further instructions, we will just have created a clone of the base class:

```python
class Rectangle(Shape):
    pass
```

```python
rect = Rectangle(6)
```

```python
rect.print_sides()
```

### In this case we want to modify the base class `Shape` to more specifically represent a rectangle
This can be achieved by:
 - overriding the parent class's `__init__()` method to initialise our `Rectangle` objects with relevant attributes, and to "hardcode" the `n_sides` attribute
 - overriding the `print_area()` method which was not implemented in `Shape`
 - extending the functionality of the parent class with a `scale()` method

```python
class Rectangle(Shape):
    name = 'rectangle'
    def __init__(self, width, height):
        Shape.__init__(self, n_sides=4)
        self.width = width
        self.height = height
        
    def scale(self, width_factor, height_factor):
        self.width = self.width * width_factor
        self.height *= height_factor
        
    def print_area(self):
        area = self.width * self.height
        print(f'This {self.name} has area {area}.')
```

```python
rect = Rectangle(5, 8)
```

```python
rect.print_sides()
```

```python
rect.print_area()
```

```python
rect.scale(2, 1)
```

```python
rect.width, rect.height
```

```python
rect.print_area()
```

### More inheritance - the Square class
This example makes use of python's `super()` function. This is a built-in function that returns a proxy object. This object allows you to refer to the parent class (superclass) without naming it explicitly. It is most commonly used in class inheritance, particularly for accessing inherited methods from the parent class in a subclass. Using `super()` in the `__init__()` method ensures that the parent class is properly initialized. It also avoids the need to hardcode the parent class name, making the code more maintainable and adaptable to changes in the class hierarchy.

In the `Square` class example below, `super()` returns a proxy object of the `Rectangle` class. This example actually restricts the functionality of `Rectangle`, so that `Square` objects cannot be initialised or scaled to become non-square.

```python
class Square(Rectangle):
    name = 'square'
    def __init__(self, side):
        super().__init__(width=side, height=side)
        
    def scale(self, factor):
        super().scale(width_factor=factor, height_factor=factor)
```

```python
sq = Square(3)
```

```python
sq.print_area()
```

```python
sq.scale(3)
```

```python
sq.print_area()
```

### The `@property` decorator
If we want the result of a method to be accessible as if it were an attribute of the object, we can use a **decorator** above the method declaration.

```python
class Square(Rectangle):
    name = 'square'
    def __init__(self, side):
        super().__init__(width=side, height=side)
        
    def scale(self, factor):
        super().scale(width_factor=factor, height_factor=factor)
        
    @property
    def area(self):
        return self.width * self.height
```

```python
sq = Square(5)

sq.area
```
