## Creating Custom Classes

Everything in Python is an object.

Every object is created from a template, or "class".

For example, a list is created from a list class, which tells a list how to be a list.

Let's take a look at creating our phone book as a Python class and then using that as an object.

A class is a blueprint, or recipe, for an object. 

### We can create a Contact class
The Contact class will hold a recipe for a Contact.
This will tell any new Contact object how to behave like a Contact.

#### Attributes
Class attributes in Python are variables defined at the class level. They are shared by all instances of the class.

The class attributes will define the pieces of data that should be stored in each Contact, e.g.
- name
- number 
- email



#### Methods
Methods in Python are functions defined within a class that operate on the instances of the class. They are an essential part of object-oriented programming in Python, allowing objects to have **behaviors** in addition to attributes. Methods encapsulate behavior that is relevant to the class, and can interact with the class's attributes (state) and other methods, encapsulating functionality within the class.

We will define the functions that will operate on that data stored, e.g. `get_initials()` to retrieve just the initials of the contact by processing their name. 


### The Process

- Create a Contact class that will act as a blueprint or recipe for creating new Contact objects
- Use the Contact class to create new contact objects
- steve will be an object of type Contact
- tamsin will be another object of type Contact

## DataFrame is an Example of a Class

### The `DataFrame` class
- The `DataFrame` class, defined in the pandas library, holds a recipe for a `DataFrame`.
- This tells any new `DataFrame` object how to behave like a `DataFrame`.

The DataFrame class defines pieces of data that should be stored in each `DataFrame`, e.g.:

- `data`: A dictionary or 2D structure to hold tabular data.
- `columns`: A list of column names in the `DataFrame`.
- `index`: A list or range of row identifiers.
- `shape`: A tuple indicating the number of rows and columns.

## Creating a Custom Contact Class

```python
class Contact:
    
    # We need to tell Python how to initialise our class
    # This is called a constructor
    def __init__(self, new_name="", new_phone="", new_email=""):
        self.name = new_name
        self.number = new_phone
        self.email = new_email
```

The above does nothing special yet, except creates the class blue print called **contact** and sets up the ability to create an **object**, a variable that will contain the set of data specific to that variable and the values assigned at initialise time.

Let's initialise a contact object called **steve**

```python
steve = Contact("Steve Shilling","555-1234",23)
```

```python
steve
```

We can see above that we have created an object called **steve** which has initialised data for name, phone and age that has been assigned to the private properties.

The output shows that we have an object in the **__main__** namespace called **Contact** that has been given a specific memory location to store that data and be capable of referencing any methods to perform actions on that data

Let's now add some methods to allow us to get the property values from the object

### View some **attributes**

Attributes are variables that need to exist for every instance of a class (e.g. steve and tamsin will both have values set for name, number and email)

```python
steve.name
```

```python
steve.number
```

### Create a **method**

Methods in Python are functions defined within a class that operate on the instances of the class. 

They are an essential part of object-oriented programming in Python, allowing objects to have **behaviors** in addition to attributes.

In other words, the Contact class defines for every contact the **behaviours** that the contact should have, e.g. the ability to retrieve just the initials of the contact.

```python
class Contact:
    
    def __init__(self, new_name="", new_phone="", new_email=""):
        self.name = new_name
        self.number = new_phone
        self.email = new_email
        
    def get_initials(self):
        parts = self.name.split()
        initials = "".join([part[0].upper() for part in parts if part])
        return initials
```

```python
steve = Contact("Steve Shilling","555-1234","steve@email.com")
```

```python
tamsin = Contact( "Tamsin Huggins","555-2149", "tamsin@email.com")
```

```python
steve.get_initials()
```

```python
tamsin.get_initials()
```

### Both variables **steve** and **tamsin** now contain unique data, but share the same **methods (functions)** to work with the data.

## From Contact to Contacts Book

We have written a Contact class to tell a contact how to be a contact. 

But what about a book of contacts?




### We can store Contact objects in a list

```python
contacts_list=[]
contacts_list.insert(0,Contact("Steve Shilling","555-1234",23))
contacts_list.insert(1, Contact("Tamsin Huggins","555-9999",21))
```

Each element of the list is now an object

```python
print(f"Tamsin's phone number is {contacts_list[1].number}")
```

### Or we could use a dictionary where the key is the person's name

```python
contacts_book={}
contacts_book["Steve Shilling"] = Contact("Steve Shilling","555-1234",23)
contacts_book["Tamsin Huggins"] = Contact("Tamsin Huggins","555-9999",21)
```

retrieve steve

```python
contacts_book["Steve Shilling"]
```

```python
print(f"Steve's phone number is {contacts_book['Steve Shilling'].number}")
```

## ContactBook Class Overview

### Attributes
- **Contacts Storage**: Employs a list or dictionary to store `Contact` objects, allowing for ordered storage or fast lookups.

### Methods
- **Add/Remove Contact**: Manages the addition or removal of contacts, enhancing data integrity.
- **Find/Print Contacts**: Facilitates efficient contact searching and display.
- **Update Contact**: Allows for updating contact details.

## Advantages of Using ContactBook Over Lists or Dictionaries

1. **Encapsulation**: Encapsulates functionality, leading to cleaner and more maintainable code. Functions that act over the contact book are defined as methods in the ContactBook class. A method like update_contact ensures any changes adhere to specific rules (such as validating email formats or ensuring no required fields are missing).
2. **Data Integrity**: Maintains data integrity by controlling access and ensuring validations. E.g. Contact can not be added if essential information is missing.
3. **Custom Behavior**: Supports custom behaviors like sorting and complex searches. For instance, sort_by_last_name might organize contacts alphabetically, while search_by_field allows searching contacts by any attribute using partial matches.

Using the `ContactBook` class provides a structured, robust approach to managing contact information, proving advantageous over basic data structures in complex scenarios.
