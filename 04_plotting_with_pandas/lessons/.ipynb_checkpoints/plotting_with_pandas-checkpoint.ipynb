# Data Visualization with `pandas.DataFrame.plot`

---

In this notebook, we’ll explore how to use [`pandas.DataFrame.plot`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html), a versatile and easy-to-use plotting interface built on top of the powerful visualization library, **Matplotlib**. Matplotlib provides the foundational plotting capabilities in Python, and `pandas.plot` builds on it, offering quick, straightforward methods for creating a variety of charts directly from data stored in `DataFrames`.

This approach is particularly useful for **quick, exploratory data visualization**. We’ll learn how to use `DataFrame.plot` to create different types of charts—such as histograms, scatter plots, and box plots—and discover how to customize these plots by adjusting arguments like colors, labels, and figure size.

While `DataFrame.plot` is excellent for rapid exploration, you can achieve **much more fine-grained control over visualizations** by combining it with direct **Matplotlib** code; learning the foundations of Matplotlib would be the next step after this notebook.

By the end of this module, you’ll be able to generate a range of informative and customised visualizations using `DataFrame.plot`.



---
## Prepare the data


import pandas package

```python


import pandas as pd
```

load the data

```python
df = pd.read_csv('../data/diamonds.csv')
```

#### Accessing Documentation for DataFrame Plotting

Use `df.plot?` to open the documentation for `pandas.DataFrame.plot`. This command provides details on available plot types, parameters, and usage examples.


```python
df.plot?
```


---
### Histogram: Diamond Price Distribution

- **Purpose**: Visualizes the distribution of a continuous variable by dividing it into intervals (bins).
  
- **Data to Plot**: `df['price']` — the `price` column from the `df` dataset.

- **Arguments**:
  - `kind`: `'hist'` (specifies a histogram plot).
  - `bins`: `30` (divides the `price` data into 30 intervals).
  - `title`: `'Distribution of Diamond Prices'` (sets the plot title).

- **Interpretation**: Useful for understanding:
  - Shape of the distribution (e.g., normal, skewed).
  - Central tendency of diamond prices.
  - Spread and range of values.

- **Example**: This histogram provides an overview of how diamond prices are distributed across different price ranges, showing areas where prices are more or less frequent.


```python
df['price'].plot(kind='hist', bins =30, title='Distribution of Diamond Prices')
```

---
### Bar Plot: Count of Diamonds by Color

- **Purpose**: Visualizes some continuous metric, usually grouped by a categorical variables. In this case, visualises the count of diamonds in each colour category.

- **Data to Plot**: `df['color'].value_counts()` — counts of each unique value in the `color` column.

- **Arguments**:
  - `kind`: `'bar'` (specifies a bar plot).

- **Interpretation**: Useful for understanding:
  - The relative frequency of each diamond color.
  - Differences in the prevalence of colors within the dataset.

- **Example**: This bar chart highlights the distribution of diamonds across color categories, allowing us to quickly compare the counts for each color.


```python
df['color'].value_counts()
```

```python
df['color'].value_counts().plot(kind='bar', title='Count of Diamonds by Color')
```

--- 

### Scatter Plot: Carat vs. Price

- **Purpose**: Visualizes the relationship between two continuous variables by plotting data points on a 2D grid.
  
- **Data to Plot**: `df` — the entire DataFrame.

- **Arguments**:
  - `kind`: `'scatter'` (specifies a scatter plot).
  - `x`: `'carat'` (sets the x-axis to the `carat` column, representing the diamond's size/weight).
  - `y`: `'price'` (sets the y-axis to the `price` column, representing the diamond's price).
  - `title`: `'Carat vs. Price'` (sets the plot title).

- **Interpretation**: Useful for understanding:
  - Trends or correlations between carat size and price.
  - Patterns, clusters, or outliers within the data.

- **Example**: This scatter plot reveals how diamond prices change with carat size, allowing us to observe any potential linear or non-linear relationships.



```python
df.plot(kind='scatter', x='carat', y='price', title='Carat vs. Price')
```

#### Categorical Box Plot: Price by Color

- **Purpose**: Compares the distribution across different categories.

- **Data to Plot**: `df` — the entire DataFrame.

- **Arguments**:
  - `column`: `'price'` (specifies the column to be plotted on the y-axis).
  - `by`: `'color'` (groups the data by the `color` column, creating a separate box plot for each color).
  - `showfliers`: `False` (hides outliers to focus on the main distribution).

- **Interpretation**: Useful for understanding:
  - The range, median, and interquartile spread of diamond prices within each color category.
  - Price variation across different colors without the influence of outlier points.

- **Example**: This box plot displays the distribution of prices across colors, helping identify variations in price spread and central tendency among different diamond colors.


```python
df.boxplot(column='price', by='color', showfliers=False)
```

## Formatting

#### Histogram with Customized Formatting

- **Purpose**: Visualizes the distribution of a continuous variable (`price`) with customized formatting options.

- **Data to Plot**: `df['price']` — the `price` column from the `df` dataset.

- **Arguments**:
  - `kind`: `'hist'` (specifies a histogram plot).
  - `bins`: `30` (divides the `price` data into 30 intervals).
  - `title`: `'Distribution of Diamond Prices'` (sets the plot title).
  - `color`: `'skyblue'` (sets the color of the histogram bars).
  - `edgecolor`: `'black'` (sets the color of the bar edges to make them distinct).
  - `figsize`: `(5, 3)` (sets the size of the figure in inches, width by height).

- **Alternative Formatting Options**:
  - **Color**: Change to other colors like `'lightgreen'`, `'orange'`, or use a color hex code (e.g., `'#FF6347'`).
  - **Figure Size**: Customize `figsize` to adjust the plot dimensions, such as `(12, 8)` for a larger plot or `(8, 4)` for a more compact view.

- **Interpretation**: This histogram provides a visually enhanced overview of the diamond prices' distribution, highlighting frequency in various price ranges.


```python
df['price'].plot(
    kind='hist',
    bins=30,
    title='Distribution of Diamond Prices',
    color='skyblue',
    edgecolor='black',
    figsize=(5, 3)
)
```

#### Scatter Plot: Carat vs. Price with Custom Formatting

- **Purpose**: Visualizes the relationship between diamond carat size and price, with custom styling for enhanced readability.

- **Data to Plot**: `df` — the entire DataFrame.

- **Arguments**:
  - `kind`: `'scatter'` (specifies a scatter plot).
  - `x`: `'carat'` (sets the x-axis to the `carat` column, representing diamond size/weight).
  - `y`: `'price'` (sets the y-axis to the `price` column, representing diamond price).
  - `title`: `'Carat vs. Price'` (sets the plot title).
  - `color`: `'purple'` (sets the color of the scatter points).
  - `alpha`: `0.3` (makes points semi-transparent to reduce overlap and improve visibility).
  - `figsize`: `(10, 6)` (defines the plot’s dimensions in inches).
  - `grid`: `True` (adds grid lines for easier reading).

- **Interpretation**: Useful for understanding:
  - How diamond prices vary with carat size.
  - Patterns, trends, or outliers, such as higher prices for larger carats.

- **Example**: This customized scatter plot reveals any relationship between carat and price, with transparency (`alpha=0.3`) to prevent overlapping points from obscuring patterns.


```python
df.plot(
    kind='scatter',
    x='carat',
    y='price',
    title='Carat vs. Price',
    color='purple',
    alpha=0.3,
    figsize=(10, 6),
    grid=True
)
```
