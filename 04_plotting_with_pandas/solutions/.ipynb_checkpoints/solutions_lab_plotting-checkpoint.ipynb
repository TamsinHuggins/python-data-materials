# Load the Data

```python
import pandas as pd
```

```python
iris_df = pd.read_csv('../data/iris.csv')
penguins_df = pd.read_csv('../data/penguins.csv')
flights_df = pd.read_csv('../data/flights.csv')
exercise_df = pd.read_csv('../data/exercise.csv')
```


---
# Iris Example Solutions

What is the distribution of petal length?

```python
iris_df['petal_length'].plot(kind='hist', title='Distribution of Petal Length')
```

How does petal length relate to petal width?

```python
iris_df.plot(kind='scatter', x='petal_length', y='petal_width', title='Petal Length vs. Petal Width')
```

What is the distrubution of petal length over just the setosa species?

```python
iris_df[iris_df['species'] == 'setosa']['petal_length'].plot(kind='hist', title='Distribution of Petal Length')
```


---
# Penguins Example Solutions

What is the distribution of penguin body mass?

```python
penguins_df['body_mass_g'].plot(kind='hist', bins=15, title='Distribution of Penguin Body Mass')
```

How does flipper length vary with body mass?

```python
penguins_df.plot(kind='scatter', x='flipper_length_mm', y='body_mass_g', title='Flipper Length vs. Body Mass')
```

How many penguins are there of each species?

```python
penguins_df['species'].value_counts().plot(kind='bar', title='Count of Each Penguin Species')
```

# Flights Example Solutions

```python
flights_df
```

How did the number of passengers each month vary over the year 1960?

```python
flights_df[flights_df['year'] == 1960].plot(kind='line', x="month", y="passengers", title='Passsengers by Month')

```

How did the number of passengers in February change between 1949 and 1960?

```python
flights_df[flights_df['month'] == "Dec"].plot(kind='line', x="year", y="passengers", title='Passsengers by Month')
```

# Exercise Example Solutions

What is the distribution of resting pulse rates?

```python
exercise_df['pulse'][exercise_df['kind'] == 'rest'].plot(kind='hist', bins=8, title='Distribution of Resting Pulse Rates')
```

 How did the pulse rate pf patient 1 (id = 1) change over time during exercise?

```python
exercise_df[exercise_df["id"] == 2].plot(kind='line', x='time', y='pulse', title='Pulse Rate Over Time')
```

How did the pulse rate pf patient 2 (id = 2) change over time during exercise?

```python
exercise_df[exercise_df["id"] == 1].plot(kind='line', x='time', y='pulse', title='Pulse Rate Over Time')
```
