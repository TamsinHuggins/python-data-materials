# Data Visualization Lab with `pandas.DataFrame.plot`

---

This lab will test your ability to use pandas.DataFrame.plot to answer questions about various datasets, practicing quick, exploratory data visualization in Python.

## The Data

Sourced from Seaborn sample datasets.

**iris**: Real measurements of iris flowers collected by Edgar Anderson; commonly used in data science for classification.

**penguins**: Real-world data on penguin species, collected from the Palmer Archipelago.

**flights**: Real data on monthly airline passenger numbers from the 1940s–1960s.

**exercise**: Simulated data on pulse rates and exercise type, useful for illustrating categorical data analysis.



## The Task

Look at the questions below. Use `pandas.DataFrame.plot` to answer questions about each dataset by creating simple plots like histograms, scatter plots, line charts, and bar charts. 

You can explore formatting options to enhance the appearance of your plots.

You will have to make use of your data filtering skills.

It is suggested that you call each of your datasets a different name, for example saving the penguins DataFrame to a variable called `penguins_df`
