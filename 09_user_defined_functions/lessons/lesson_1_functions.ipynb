# User-Defined Functions

---

In Python, a user-defined function is a block of organized, reusable code that is used to perform a single, related action. This allows users to create custom functionality beyond built-in functions, giving them flexibility in their coding projects. Here are just a few reasons that you may want to consider defining functions in your code:

 - **Reusability**: If you find yourself writing the same code in multiple places, encapsulating it in a function can allow for reuse, reducing redundancy.
 - **Modularity**: Functions allow you to break down complex tasks into smaller, more manageable parts, improving readability and maintainability.
 - **Abstraction**: Functions can abstract away the implementation details. The user doesn't need to know how a function works internally; they only need to know its inputs and expected outputs.
 - **Code Organization**: Grouping related code into functions makes your codebase more organized and easier to navigate.
 - **Reduction in Code Size**: By eliminating repetitive code through the use of functions, the overall size of your code can be significantly reduced.


A function is a set of instructions to perform a specific task.
- Python functions use round brackets.
- Functions often take some input and perform an operation based on that input.
- The input that we pass into the function is often called the argument.
- The argument is passed into the function inside the round brackets

### Indentation is important in Python
In many programming languages, curly braces { } are used to define a block of code, such as a function body or a loop. In contrast, Python uses indentation (whitespace) to determine the **scope** of blocks of code. This means that the amount of lateral whitespace (typically spaces or tabs) at the start of a line plays a crucial role in the structure and semantics of a Python program.

Consistent indentation is not just a matter of style in Python - it's syntactically significant. Lines of code with the same amount of indentation are considered to be in the same "block". This is irrespective of the amount of vertical space (line breaks) between them.

```python
def message():
    print("This message is only printed as a result of calling the function.") # this line is automatically indented by four spaces
    print("This is the message") # the last line of code at this indentation level signifies the end of the function

print("This print statement is not inside the message function and gets printed first", "\n")

message()
```

## Function Arguments

 Python supports multiple types of arguments:
- Positional Arguments
- Keyword Arguments
- Default Arguments (set in the function definition)

Positional Arguments

```python
# Correct arguments order
job_info('Phoebe', 'developer')
# Wrong arguments order
job_info('developer', 'Pheobe')
```

### Keyword Arguments
- you can pass arguments using the names of their corresponding parameters.
- in this case, the order of the arguments no longer matters.

```python


job_info(job='developer', name='Phoebe')

job_info(name='Phoebe', job='developer')

```

 
### Parameters with Default Arguments
 - You can specify default values for arguments when defining a function. 
 - The default value is used IF the function is called without a corresponding argument.


```python


def job_info_default_dev(name, job='developer'):
    print(name, 'is a', job)

job_info_default_dev('Phoebe')
job_info_default_dev('Phoebe', 'data analyst')

```

## Variable Scope
In Python, variables have either a local scope or a global scope:
 - Local Scope: Variables defined inside a function are said to have a **local** scope, and they cannot be accessed outside of that function
 - Global Scope: Variables defined outside of functions have a **global** scope and can be accessed and modified from any function

If a variable is defined at **both** levels, the local variable takes precedence inside the function. This is known as variable **shadowing**.

```python
x = 10  # This is a global variable

def example_function():
    x = 5  # This is a local variable
    print(f"Inside function: x = {x}")

example_function()

print(f"Outside function: x = {x}")
```

## The Return Statement

- To return a value from a function,  use a return statement. 
- Once a return statement is executed, nothing else in the function body is executed.
- If you do not include any return statement, the function automatically returns None.



```python

# so far, we our functions have had no return statements so return None.
return_result = job_info("Phoebe", "developer")
print(return_result)

```

## Returning More than One Value
Many functions return more than one value. By default, if you send more than one value out via the return statement, these values get packed into a `tuple`. It is often useful to "unpack" the values retured by a function into individual variables, for further use in your program.

```python
def powers(x):
    squared = x ** 2
    cubed = x ** 3
    return squared, cubed
```

```python
powers(7)
```

```python
# this will "unpack" the tuple
sqr, cub = powers(7)
```
