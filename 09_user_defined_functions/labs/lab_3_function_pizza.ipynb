## Starter Code, Uneditable

```python
PRICES = {
    "pepperoni": {9: 8}
    # add other flavours and their prices to the prices dictionary!
}


def order_pizza():
    flavour = handle_flavour()
    if flavour:
        size = handle_size(flavour)
        if size:
            cost = PRICES[flavour][size]
            print(f'you have ordered a {size}" {flavour} pizza. That will cost £{cost}.')
            
        else: 
            return None
    else:
        return None
    
    
def handle_flavour():
    flavour = input('Would you like Pepperoni, Margarita or Veg-Feast?: ').lower()
    if flavour in PRICES.keys():
        print(f"{flavour} is available")
        return flavour       
    else:
        print('not a valid flavour, try again')
        return None 
    
    
    
def handle_size(chosen_flavour):
    try:
        size = int(input(f'Please choose between our 9", 11" or 13" pizza. Type 9, 11 or 13: '))
        if size in PRICES[chosen_flavour].keys():
            print(f'you selected {size}"')
            return size
        else:

            print("please choose a valid size")
            return None  
    except:
        print("You must enter a number")
        return None    
        
    
order_pizza()
```
