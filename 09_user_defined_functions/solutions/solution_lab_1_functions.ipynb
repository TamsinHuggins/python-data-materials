#### Question 1:
Define a function `sum_two_numbers` that takes two positional arguments and returns their sum.

```python
def sum_two_numbers(a, b):
    return a + b

sum_two_numbers(4, 6)
```

#### Question 2:
Define a function `find_extremes` that takes a list or tuple of numbers as a single argument and returns the minimum and maximum numbers from this collection.

```python
def find_extremes(seq):
    return min(seq), max(seq)

seq = [4, 2, 8, 9, -11, 1]

find_extremes(seq) # should return (-11, 9)
```
