###  <span style="color:green"> Function Challenge </span>

In some cells below:

1. Define a function called circle_area that calculates the area of a circle ( pi times radius squared)

- Function inputs: radius of a circle
- Function outputs: area of the circle

2. make the function print out "calculating the area of a circle with a radius of {x}", replacing x with the number recieved from the argument.

3. make the function return the result of the calculation
 
4. Call the function and save the result to a variable

5. print the variable and check the result.

Bonus: set the default radius to be 10

```python

import math

def circle_area(radius=10):
    # calculate area
    print(f"calculating the area of circle with radius {radius}")
    area = math.pi * (radius ** 2)
    return area


result = circle_area(30)
print(result)

print(circle_area())

```
