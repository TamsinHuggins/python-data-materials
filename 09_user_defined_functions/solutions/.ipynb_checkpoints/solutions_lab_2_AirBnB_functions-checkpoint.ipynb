
### 3b) Filter Listings by Amenity

#### Objective
Create a function named `get_listings_with_amenity` that receives the name of an amenity as input and prints out all listings that include that amenity.

#### Define the Function 
Write a function called `get_listings_with_amenity`. This function should take one parameter, `amenity`, which is a string that represents the amenity (e.g. Pool table, Gym, Coffee maker) that you want to check for in each listing.

#### Accessing Data
Assume you have a global list `data` available, where each element is a dictionary representing a property listing with attributes such as `'name'` and `'amenities'`.

#### Implement the Filtering Logic
- **Iterate Through Listings**: Use a `for` loop to go through each listing in the `data`.
- **Check for Amenity**: Inside the loop, use an `if` statement to check if the specified `amenity` is in the listing's `'amenities'` list.
- **Output Listing Names**: If the listing contains the amenity, print the listing’s name.

#### Test the Function
- **Initial Test**: Call `get_listings_with_amenity` with the argument `'Pool table'` to find all listings that have a pool table.
- **Further Testing**: Call the function again with the argument `'Gym'` to filter for listings with a gym.


 Test your function by calling it with the argument 'Pool table'
 
 Test again your function by calling it with the argument 'Gym'


````python
def get_listings_with_amenity(amenity):
    print(f"Listings with {amenity}:")
    for listing in data:
        if amenity in listing['amenities']:
            print(listing['name'])
    

get_listings_with_amenity("Gym")
````

## 4) Show the Price in Multiple Currencies
Create a function named `show_currencies` that receives a listing object containing a price in USD and prints the price converted into Euros and Indian Rupees.

### Define the Function:
Write a function called `show_currencies` that takes one parameter, `listing`, which is a dictionary containing at least the key `'price'`.

### Extract and Clean Price:
Extract the price from the listing and remove any currency symbols (e.g., '$'). Hint: are there any string methods that can replace a character with something else?

### Convert and Calculate:
- Convert the cleaned price from a string to a float.
- Define exchange rates for converting USD to Euros and Rupees.
- Calculate the converted prices. You may wish to round them to 2 decimal places.

### Print Converted Prices:
Print the original price and the converted prices with appropriate currency symbols.


### Try the function
 Check that your function works. Pass in some listings, e.g.

```
show_currencies(data[2])
show_currencies(data[3])


```

````python
def show_currencies(listing):
    # Assume 'price' is a key in the 'listing' dictionary that gives us the price as a string with a dollar sign
    price = listing['price']
    
    # Remove the dollar sign from the price string
    clean_price = price.replace('$', '')
    
    # Convert the cleaned price from string to float for calculations
    numeric_price = float(clean_price)
    
    # Define exchange rates; you might need to update these based on current rates
    usd_to_euro = 0.91  # Example rate, 1 USD = 0.91 EUR
    usd_to_rupee = 82.50  # Example rate, 1 USD = 82.50 INR
    
    # Calculate and round the price in Euros
    euro_price = round(numeric_price * usd_to_euro, 2)
    
    # Calculate and round the price in Indian Rupees
    rupee_price = round(numeric_price * usd_to_rupee, 2)
    
    # Round the USD price
    numeric_price = round(numeric_price, 2)
    
    # Print the prices in different currencies
    print(f"Price in USD: ${numeric_price}")
    print(f"Price in Euros: €{euro_price}")
    print(f"Price in Indian Rupees: ₹{rupee_price}")

````

````python
show_currencies(data[2])
````

````python
show_currencies(data[3])
````
