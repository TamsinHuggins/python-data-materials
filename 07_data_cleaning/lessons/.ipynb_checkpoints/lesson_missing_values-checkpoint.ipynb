## Detect missing data

Pandas includes a number of functions to detect missing or invalid data.

- isnull  - Returns a Series containing True/False indicating if each value is missing.
- notnull - Opposite (negation) of isnull: True if value is not null, False otherwise.
- sum - how many null or not nulls exist


```python
df.isnull()
df.notnull()
df.isna()
```

```python

# How many in each column
df.isna().sum()

# How amny are NaN in the entire dataset
df.isna().sum().sum()
```

# Dealing with Missing Values

What can we do with NaNs?

 - drop (remove) rows
 - drop (remove) columns
 - fill NaNs in with hard-coded values
 - fill NaNs in with computer-caluclated values (e.g. interpolate)
 - 
 


## Drop Na

- **dropna()**  - Will detect and remove rows or columns (it's usually used for rows) where data is missing.

- Returns a copy, not the original.

- Catch result in a new variable OR set **inplace=True** to alter the original DataFrame.

Drop rows with missing vals

```python
df.dropna() 

#df.dropna(inplace=True)
```

Drop columns with missing vals

```python
df.dropna(axis=1) #inplace=True
```

## Fill Na

Some methods will throw errors when applied to data with NaNs. In some contexts, filling all NaN values with a single value such as 0 or the mean of a column is suitable.

```python
# use fillna - returns a new object, can use inplace=True if desired
pop_fillna = df.copy(deep=True)

pop_fillna.fillna(0, inplace=True)
pop_fillna
```

#### Fill Forward

Forward fill: fill horizontally left to right, this does not save changes to the DataFrame.

Can you spot an issue with the ASM row?

```python

df.fillna(method='ffill', axis=1)
```

#### Fill backward
fill horizontally right to left, this does not save changes to the DataFrame.

```python
df.fillna(method='bfill', axis=1)
```

##  Interpolate



Find the mid point between the values on either side.

```python
df.iloc[:, 4:12].interpolate(axis=1)
```

Save over the section with the interpolated version

```python
df.iloc[:, 4:12] = df.iloc[:, 4:12].interpolate(axis=1)
```

American Samoa can't be interpolated because its NaN value sits on the end.

```python
df.isna().any(axis=1)
```

```python
backfilled = df.fillna(method='bfill', axis=1)
```

Check that we have really dealt with the NaNs

```python
backfilled.isna().any()
```

## Scikit Learn imputers 

`SimpleImputer` is form the scikit learn machine learning libraries

the following cells show how to employ various strategies to impute data - `mean`, `most_frequent` and `constant`

### strategy = 'mean'

```python
from sklearn.impute import SimpleImputer

df_houses = pd.read_excel('../Data/data_sets.xlsx', sheet_name='HouseData')
df_houses
```

```python

imputer = SimpleImputer(missing_values = np.nan, 
                        strategy = 'mean', verbose=0)

cols = ['Age', 'Salary']

imputer = imputer.fit(df_houses[cols])

df_houses[cols] = imputer.transform(df_houses[cols])

df_houses
```

### strategy = 'most_frequent'

```python
# How many of each value are in purchased column
display (df_houses['Purchased'].value_counts())

cols = ['Purchased']
imputer = SimpleImputer(missing_values = np.nan, 
                        strategy = 'most_frequent', verbose=0)
imputer = imputer.fit(df_houses[cols])

df_houses[cols] = imputer.transform(df_houses[cols])
```

### strategy = 'constant'

- need a fill value for this approach

```python
cols = ['Country']
imputer = SimpleImputer(missing_values = np.nan, 
                        strategy = 'constant', fill_value='France', verbose=0)

imputer = imputer.fit(df_houses[cols])

df_houses["Country"] = imputer.transform(df_houses[cols])
```

```python
df_houses
```

## Replacing Values
- **df.replace(to_replace, value)** : find and replace specific values.
- The parameters **to_replace** and **value** can both be either single values or lists of values.
- Returns a copy so again either use **inplace=True** OR catch the returned DataFrame in a new variable.

```python
df_missing = pd.read_excel('../Data/data_sets.xlsx', sheet_name='MissingData')
df_missing.fillna(0, inplace=True)
df_missing
```

```python
df_missing.replace(2,22, inplace=True)
df_missing
```

```python
df_missing = df_missing.replace(['I', 'III'],[11, 33])
df_missing
```

```python
# or use variables
orig_vals = ['I', 'III']
new_vals = [11, 33]

df_missing.replace(orig_vals,new_vals, inplace=True)
df_missing
```
