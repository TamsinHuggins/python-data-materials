# Data Cleaning with Pandas



- Dealing with duplicates
- Dropping rows
- Droppong columns
- Renaming columns
- Setting an index


```python
import pandas as pd
```

```python
df = pd.read_csv('../Data/world_population_dirty.csv')
```

## Dealing with duplicates

.duplicated() shows us whichever rows are duplicated

```python
df[df.duplicated()]
```

Do we have Argentina twice in the data? Let's search.

```python
df[df['CCA3'] == 'ARG']
```

#### Remove the chosen row using a standard drop method.
Note that this will not make a change to the DataFrame without setting inplace to true 

```python
df.drop(234) # df.drop(234,inplce=True)
```

#### Remove using drop duplicates

```python
print(df.shape)
df.drop_duplicates(inplace=True)
print(df.shape)
df
```

## Dropping a column

let's say we did a column calculation wrong and appended it to the DataFrame, then wanted to remove it

```python
df['accidental column'] = 'whoops!'
df
```

axis=1 means drop from the columns, not the rows. inplace=True to make the change to the DataFrame.

```python
df.drop('accidental column', axis=1) #df.drop('accidental column', axis=1, inplace=True)
```

.pop is an INPLACE method so the column will be removed.

```python
df.pop('accidental column')
```

```python
df
```

## Renaming columns

Rename World Population Percentage to something shorter

```python
df.rename(columns={'World Population Percentage': '%_of_world'}) #inplace = True
```

Rename multiple at a time.

```python
df.rename(columns={'2022 Population': '2022', 
                         '2020 Population': '2023'}) #inplace = True
```

## Setting/changing an index

Demo the change

```python
df.set_index('CCA3')
```

Implement the change to the dataframe

```python
df.set_index('CCA3', inplace=True)
```

We can now select rows of the data using loc with index names (not just iloc with index numbers)

```python
df.loc['AND']
```

```python
df.loc['ALB']
```

```python
df.loc[['ALB', 'AND']]
```
