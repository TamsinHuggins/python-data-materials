## What is NaN?

In Python, **NaN** stands for "Not a Number." It is a special floating-point value defined in the `numpy` library to represent missing, undefined, or unrepresentable numeric data. NaN is commonly encountered in datasets when:
- A value is missing.
- A calculation results in an undefined value (e.g., `0/0` or the square root of a negative number).

NaN behaves uniquely:
- It is not equal to any other value, including itself.
- It is often used as a placeholder in numerical computations and data analysis.



NaN is:
- Not A Number
- Different from all other data types
- Not equal to anything

- Defined in `numpy`

### Comparing NaN to other values

NaN is not equal to True, False, 0 or Empty Strings

```python
from numpy import NaN 

print(NaN == True)

print(NaN == False)

print(NaN == 0)

print(NaN == '')
```

## Detect missing data

Pandas includes a number of functions to detect missing or invalid data.
The `df.isna()` method is used to detect missing values (NaN) in the DataFrame. By default, it returns a DataFrame of the same shape as `df`, with `True` indicating a missing value and `False` otherwise. 


```python

df.isna()
```

- `.notna()` - inverse of isna

```python

df.notna()
```

---

## Chaining `isna()` with Other Functions 

`.isna()` gives us a True/False table, highlighing NaN values with `True` (which is equivalent to `1`) and `False` (which is equivalent to `0`). We can then perform functions over this table True/False DataFrame to summarise and aggregate.




#### `.any()` For identifying the locations of missing values

#### Identify **columns** with missing values

The `df.isna()` method is used to detect missing values (NaN) in the DataFrame. 

```python
df.isna().any()
```

#### Identify **rows** that contain missing values

The `df.isna()` method is used to detect missing values (NaN) in the DataFrame. 

The `.any()` method is a boolean aggregation function that checks whether **at least one** value in the specified axis is `True`. 

By setting `axis=1`, we instruct Pandas to perform this aggregation across the **columns** for each row. As a result, this will return a Series of boolean values, where `True` indicates that the corresponding row contains **at least one missing value**, and `False` indicates that the row contains no missing values.

```python
df.isna().any(axis=1)
```

Show rows that contain missing values

```python
df[df.isna().any(axis=1)]
```

Identify columns containing missing values

By comparning the shape of the column with its value counts we can also see if there are missing values


#### `.sum()`

- This function calculates the sum along a specified axis. By default, it sums along each column (`axis=0`).
- When applied to the boolean DataFrame returned by `.isna()`, it treats `True` as `1` and `False` as `0`.
- The result is a Series where each value represents the count of missing values in the corresponding column.


```python
# How many in each column
df.isna().sum()
```

### Summing missing data over the rows

If you want to sum the missing values over the rows instead of the columns, you can set `axis=1` to flip the direction in which the calculation is 
performed

Here we see that AND (the row representing Andorra) has a missing value.

```python
df.isna().sum(axis=1)
```

#### Calculate total missing values over the DataFrame
`df.isna().sum().sum()`

- **Step 1: `df.isna()`**
  - This checks for missing values (NaN) in the DataFrame.
  - It returns a DataFrame of the same shape where each cell is `True` if the value is missing, otherwise `False`.

- **Step 2: `.sum()` (First Sum)**
  - This calculates the sum along each column (`axis=0`) of the boolean DataFrame returned by `.isna()`.
  - Since `True` is treated as `1` and `False` as `0`, this results in a Series where each value represents the count of missing values in the corresponding column.

- **Step 3: `.sum()` (Second Sum)**
  - This calculates the sum of the Series obtained in Step 2.
  - The result is a single number representing the total count of missing values across all columns and rows of the original DataFrame.


```python
df.isna().sum().sum()
```

## Dealing with Missing Values

What can we do with NaNs?

 - drop (remove) rows
 - drop (remove) columns
 - fill NaNs in with hard-coded values
 - fill NaNs in with computer-caluclated values (e.g. interpolate)
 - 
 


### Remove missing values with`.dropna()`

- **dropna()**  - Will detect and remove rows or columns (it's usually used for rows) where data is missing.

- Returns a copy, not the original.

- Catch result in a new variable OR set **inplace=True** to alter the original DataFrame.

### Drop rows with missing values

```python
df.dropna()  #inplace=True
```

### Drop columns with missing values

- You can set `axis=1` to drop columns instead of rows


```python
df.dropna(axis=1) #inplace=True
```


---
### Fill missing values with `.fillna()`

- In this example, we fill all missing values with `0`
- What to fill missing values with depends on the task at hand

```python
df.fillna(0) #inplace=True
df
```

#### Fill Forward

- `method="ffill"`

Can you spot an issue with the ASM row?

```python

df.fillna(method='ffill')
```

#### Fill backward
fill horizontally right to left, this does not save changes to the DataFrame.

```python
df.fillna(method='bfill', axis=1)
```

---
###  Fill missing values with `.interpolate()`



Find the mid point between the values on either side.

```python
df.iloc[:, 4:12].interpolate(axis=1)
```

Save over the section with the interpolated version

```python
df.iloc[:, 4:12] = df.iloc[:, 4:12].interpolate(axis=1)
```

American Samoa can't be interpolated because its NaN value sits on the end.

```python
df.isna().any(axis=1)
```

```python
backfilled = df.fillna(method='bfill', axis=1)
```

Check that we have really dealt with the NaNs

```python
backfilled.isna().any()
```
