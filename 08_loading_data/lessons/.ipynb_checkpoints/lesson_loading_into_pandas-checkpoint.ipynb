# Loading Data into Pandas

We have seen that Pandas is a powerful library for working with structured data in Python.

### Why is Pandas powerful?
- Seamlessly integrates with Python's data ecosystem (e.g., NumPy, Matplotlib, Scikit-learn).
- Provides the flexible and intuitive **DataFrame** object for tabular data.
- Supports multiple data formats: CSV, Excel, JSON, SQL, and APIs.
- Includes powerful tools for filtering, aggregation, reshaping, and joining datasets.
- Handles missing data efficiently and enables quick data cleaning.
- Facilitates time-series analysis and date/time handling.
- Optimized for performance with vectorized operations.

### Why load data into Pandas?
- **Exploration**: Quickly inspect and summarize datasets.
- **Cleaning**: Handle missing, inconsistent, or erroneous data.
- **Transformation**: Prepare data for analysis or machine learning pipelines.
- **Transparency**: Code-based workflows in Pandas provide a clear, auditable history of data manipulations.
- **Analysis**: Perform calculations, group-by operations, and create pivot tables.
- **Export**: Save processed data to various formats (e.g., CSV, Excel, databases).
- **Efficiency**: Simplify complex data workflows and ensure reproducibility.


### File Paths

### `.` (Single Dot)
- Refers to the **current directory**.
- Use it when your file is in the same directory as your script or notebook.
- Example: 
  ```python
  df = pd.read_csv('./data.csv')  # File in the current directory


### `..` (Double Dot)
- Refers to the **parent directory** (one level up).
- Use it to access files located outside the current directory.

Example:
```python
df = pd.read_csv('../file.csv')  # File in the parent directory
```

### `/` (Slash)
- Indicates entering a directory in the file path.
- Use it to navigate through folders within your directory structure.
Examples:

- Navigate to a subdirectory:
  ```python
  df = pd.read_csv('./subfolder/data.csv')  # File in a subfolder



### Examples

| Path                          | Meaning                                          |
|-------------------------------|--------------------------------------------------|
| `./file.csv`                  | File in the current directory                   |
| `./folder/file.csv`           | File in a subfolder of the current directory    |
| `../file.csv`                 | File in the parent directory                    |
| `../../file.csv`              | File two levels up in the directory hierarchy   |
| `../folder/file.csv`| File in a a sibling directory      |

## Reading from Excel

Loading data into Pandas from an Excel file (`.xlsx`) differs from loading from a CSV in a few key ways:

- **Function**: Use `pd.read_excel()` for Excel files, whereas `pd.read_csv()` is used for CSV files.
- **Sheet Selection**: Excel files can contain multiple sheets; specify which one to load using the `sheet_name` parameter.



### File Path:
- `"../data/all_data.xlsx"` specifies the relative path to the Excel file.
- The `..` indicates moving up one level in the file structure from the current working directory, followed by navigating to the `data` folder and accessing the `all_data.xlsx` file.

### Sheet Selection:
- `sheet_name="planets"` specifies which sheet to load from the Excel file.
- Excel files can contain multiple sheets, and this parameter ensures only the "planets" sheet is loaded into the DataFrame.

### Result:
- The `planets_df` variable now holds the data from the "planets" sheet in the specified Excel file as a Pandas DataFrame.
- You can use this DataFrame for further analysis, such as exploring data, creating visualizations, or performing calculations.


````python
planets_df = pd.read_excel("../data/all_data.xlsx", sheet_name="planets")
````
