### File Paths

### `.` (Single Dot)
- Refers to the **current directory**.
- Use it when your file is in the same directory as your script or notebook.
- Example: 
  ```python
  df = pd.read_csv('./data.csv')  # File in the current directory


### `..` (Double Dot)
- Refers to the **parent directory** (one level up).
- Use it to access files located outside the current directory.

Example:
```python
df = pd.read_csv('../file.csv')  # File in the parent directory
```

### `/` (Slash)
- Indicates entering a directory in the file path.
- Use it to navigate through folders within your directory structure.
Examples:

- Navigate to a subdirectory:
  ```python
  df = pd.read_csv('./subfolder/data.csv')  # File in a subfolder



### Examples

| Path                          | Meaning                                          |
|-------------------------------|--------------------------------------------------|
| `./file.csv`                  | File in the current directory                   |
| `./folder/file.csv`           | File in a subfolder of the current directory    |
| `../file.csv`                 | File in the parent directory                    |
| `../../file.csv`              | File two levels up in the directory hierarchy   |
| `../folder/file.csv`| File in a a sibling directory      |
